## 说明

C3 的评测是两件事：Vina 对接算结合亲和力，加一个 KRAS G12D 活性模型算活性。

**Vina 有了** —— 小分子 RL 那条线一直在用，aarch64 版本装好了，跑得动。

**活性模型没有。** `best_g12d_model.joblib` 不随仓库发布，
得直接找论文作者要。没有它，活性这一维就没法算，
超体积也就无从谈起 —— 而超体积正是论文 C3 报的那个数。

## 能不能自己训一个

理论上可以：我们在小分子 RL 那条线上确实训过 G12D 和 G12C 的 QSAR 模型
（`ldm_rl/results/g12d_qsar_matched_*/best_model.joblib`）。

**但那样就不是复现了。** 超体积对活性模型的标定极其敏感 ——
换一个模型，Pareto 前沿的形状跟着变，+62.4% 这个数就没有可比性。
自己训一个来「复现」，量的是我们自己的模型，不是论文的结论。

所以这条要么拿到原始的 joblib，要么就不做。**中间没有便宜的路。**
